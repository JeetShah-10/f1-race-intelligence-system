# FastF1 Data Probe: Bahrain 2023

**Objective:** Validate that FastF1 provides the necessary features for our ML pipeline.
**Target Features:**
- Lap Times (Pace)
- Tyre Life & Compound
- Track Status (Safety Car events)
- Telemetry completeness

```python
import fastf1
import pandas as pd

# Enable Cache
fastf1.Cache.enable_cache('../backend/cache/fastf1')

# Load Session
session = fastf1.get_session(2023, 'Bahrain', 'R')
session.load()
```

## 1. Lap Data Integrety

```python
# Inspect Columns
print("Available Columns:", session.laps.columns.tolist())

# Check Max Verstappen's Laps
ver_laps = session.laps.pick_driver('VER')
print(ver_laps[['LapNumber', 'LapTime', 'TyreLife', 'Compound', 'TrackStatus']].head(10))
```

## 2. Safety Car Detection
Track Status '1' = Green. Anything else implies flags.

```python
sc_laps = session.laps[session.laps['TrackStatus'] != '1']
print(f"Non-Green Laps: {len(sc_laps)}")
print(sc_laps[['LapNumber', 'TrackStatus']].head())
```
